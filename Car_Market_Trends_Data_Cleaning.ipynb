# Data Cleaning & Preprocessing Report
## Car Market Trends Analysis (Car Dekho Dataset)

This notebook contains the complete, reproducible workflow for acquiring, exploring, auditing, cleaning, and preprocessing the Car Dekho used car market dataset.

### 1. Import Libraries & Load Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_theme(style="whitegrid")

# Load dataset
df = pd.read_csv("1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv")
print("Initial Dataset Shape:", df.shape)
df.head()
```

### 2. Initial Data Exploration & Inspection

```python
# Summary Information & Data Types
print("--- Data Schema ---")
df.info()

print("\n--- Statistical Summary ---")
df.describe(include='all')
```

### 3. Data Quality Audit & Cleaning

```python
# Check Missing Values
print("Missing Values per Column:")
print(df.isnull().sum())

# Check & Handle Duplicates
duplicates_count = df.duplicated().sum()
print(f"\nDuplicate rows found: {duplicates_count}")

# Drop Duplicates
df_cleaned = df.drop_duplicates().copy()
print("Dataset Shape after removing duplicates:", df_cleaned.shape)
```

### 4. Outlier Inspection & Visualizations

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Selling Price & Present Price Boxplot
sns.boxplot(data=df_cleaned[['Selling_Price', 'Present_Price']], ax=axes[0])
axes[0].set_title("Boxplot of Selling Price vs Present Price (Lakhs)")
axes[0].set_ylabel("Price (in Lakhs)")

# Distribution of Kilometers Driven
sns.histplot(df_cleaned['Kms_Driven'], kde=True, bins=30, color='skyblue', ax=axes[1])
axes[1].set_title("Distribution of Kilometers Driven")
axes[1].set_xlabel("Kilometers Driven")

plt.tight_layout()
plt.show()
```

### 5. Feature Engineering & Preprocessing

```python
# 1. Feature Engineering: Calculate Car Age & Price Depreciation
current_year = 2026
df_cleaned['Car_Age'] = current_year - df_cleaned['Year']
df_cleaned['Price_Depreciation'] = df_cleaned['Present_Price'] - df_cleaned['Selling_Price']
df_cleaned['Depreciation_Pct'] = (df_cleaned['Price_Depreciation'] / df_cleaned['Present_Price']) * 100

# 2. One-Hot Encoding for Categorical Attributes
df_encoded = pd.get_dummies(
    df_cleaned, 
    columns=['Fuel_Type', 'Seller_Type', 'Transmission'], 
    drop_first=True
)

# 3. Model-Ready Dataset (drop identifier/raw temporal columns)
df_model_ready = df_encoded.drop(columns=['Car_Name', 'Year'])

print("Preprocessed Data Shape:", df_model_ready.shape)
df_model_ready.head()
```

### 6. Correlation Heatmap Analysis

```python
numeric_cols = ['Selling_Price', 'Present_Price', 'Kms_Driven', 'Car_Age', 'Owner']
corr = df_cleaned[numeric_cols].corr()

plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix of Numeric Features")
plt.show()
```
